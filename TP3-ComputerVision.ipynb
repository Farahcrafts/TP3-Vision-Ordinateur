## Exercice1: Analyse d'une image

```python
import cv2

#Lire une image 
img = cv2.imread('images/cat.jpg')

#Afficher l'image dans une fenetre 
cv2.imshow('Image originale', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Conversion en niveaux de gris 
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow('Imae en niveaux de gris', gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Detection des contours avec l'algo de Canny 
edges = cv2.Canny(img, 100, 200)
cv2.imshow('Contours detectes', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

1. La matrice retournee par OpenCV pour l'image représente la grille de pixels de l'image. Chaque chiffre correspond à l'intensité lumineuse (et la couleur) d'un pixel.

2. **Pourquoi convertir en gris ?** Pour simplifier les données (1 canal au lieu de 3), ce qui rend les calculs beaucoup plus rapides et facilite la détection des formes et des contours sans se soucier des couleurs.

## Exercice 2: Lecture video / webcam

```python
cap = cv2.VideoCapture('images/video.mp4')
while True:
    ret, frame = cap.read()
    if not ret:
        break
    cv2.imshow('Webcam', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
from ultralytics import YOLO
import cv2

# 1. Charger le modèle de segmentation YOLOv8 (version 'nano' pour la rapidité)
model = YOLO('yolov8n-seg.pt')

# 2. Charger votre image
img_path = "images/manyobjects.jpg"
img = cv2.imread(img_path)

# 3. Lancer la prédiction
results = model(img)

# 4. Afficher le résultat (la fonction plot() dessine déjà les masques et les étiquettes)
annotated_img = results[0].plot()

cv2.imshow("YOLOv8 Segmentation", annotated_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Exercice 4:

```python
import dlib
import cv2

# Initialiser le détecteur
detector = dlib.get_frontal_face_detector()

# Lire l'image
img = cv2.imread("images/manypeople.jpeg")

# Convertir en gris
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# LE CHANGEMENT EST ICI : on ajoute ", 1" pour agrandir l'image 
# et aider Dlib à trouver les petits visages.
faces = detector(gray, 1) 

# Petite vérification pour être sûr
print(f"Nombre de visages trouvés : {len(faces)}")

# Dessiner les rectangles
for face in faces:
    x, y, w, h = face.left(), face.top(), face.width(), face.height()
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)

# Afficher (sans accents dans le titre pour éviter les symboles bizarres)
cv2.imshow("Visages detectes", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
